```python
import re

board = ["1", "2", "3",
         "4", "5", "6",
         "7", "8", "9"]

def display():
    print()
    print(board[0], "|", board[1], "|", board[2])
    print("--+---+--")
    print(board[3], "|", board[4], "|", board[5])
    print("--+---+--")
    print(board[6], "|", board[7], "|", board[8])
    print()

def check_winner():

    if board[0] == board[1] == board[2]:
        return board[0]
    if board[3] == board[4] == board[5]:
        return board[3]
    if board[6] == board[7] == board[8]:
        return board[6]


    if board[0] == board[3] == board[6]:
        return board[0]
    if board[1] == board[4] == board[7]:
        return board[1]
    if board[2] == board[5] == board[8]:
        return board[2]


    if board[0] == board[4] == board[8]:
        return board[0]
    if board[2] == board[4] == board[6]:
        return board[2]

    return None

def is_board_full():
    for cell in board:
        if cell not in ["X", "O"]:
            return False
    return True

def computer_move():



    for i in range(9):
        if board[i] not in ["X", "O"]:
            old = board[i]
            board[i] = "O"

            if check_winner() == "O":
                return

            board[i] = old


    for i in range(9):
        if board[i] not in ["X", "O"]:
            old = board[i]
            board[i] = "X"

            if check_winner() == "X":
                board[i] = "O"
                return

            board[i] = old


    if board[4] == "5":
        board[4] = "O"
        return


    for i in [0, 2, 6, 8]:
        if board[i] not in ["X", "O"]:
            board[i] = "O"
            return


    for i in range(9):
        if board[i] not in ["X", "O"]:
            board[i] = "O"
            return



print("TIC-TAC-TOE")
print("You = X")
print("Computer = O")

for turn in range(9):

    display()


    position = int(input("Enter position (1-9): ")) - 1

    if position < 0 or position > 8:
        print("Invalid position!")
        continue

    if board[position] in ["X", "O"]:
        print("Position already occupied!")
        continue

    board[position] = "X"


    winner = check_winner()

    if winner == "X":
        display()
        print("You Win!")
        break

    if winner is None and is_board_full():
        display()
        print("Game Draw!")
        break

    if winner is None and not is_board_full():
        computer_move()

        winner = check_winner()

        if winner == "O":
            display()
            print("Computer Wins!")
            break

        if winner is None and is_board_full():
            display()
            print("Game Draw!")
            break
    else:
        pass
```
